#### Run this cell to connect to your GIS and get started:

```python
import arcgis
from arcgis.gis import GIS
import requests
import pandas as pd

gis = GIS("home")
```

#### This code downloads XML parking data from Bonn, converts it to a pandas DataFrame, and calculates the occupancy percentage from the "frei" (free) and "gesamt" (total) columns.

```python
url = 'https://www.bcp-bonn.de/stellplatz/bcpext.xml'

response = requests.get(url)
xml = response.text


df = pd.read_xml(url,parser="etree")

df["occupied"] = (1-df["frei"].astype(float)/df["gesamt"].astype(float))*100
```

#### This code retrieves a specific ArcGIS Online item by its ID and accesses the first table within that item for data upload.

```python
portal_item = gis.content.get('a3a5408a567d43d0a04e80333ae787b6')
portal_table = portal_item.tables[0]
portal_item
```

#### This converts the pandas DataFrame into an ArcGIS FeatureSet format that can be uploaded to ArcGIS Online tables.

```python
df_fs = arcgis.features.FeatureSet.from_dataframe(df) 
```

#### This uploads the FeatureSet data to the ArcGIS Online table by adding new records.

```python
portal_table.edit_features(adds = df_fs)
```
